```python
model1 = YOLO("/content/model1.pt")
model4 = YOLO("/content/model4.pt")

print("Model 1:", model1.ckpt["train_args"]["name"])
print("Model 4:", model4.ckpt["train_args"]["name"])
```

Validation at 640

```python
metrics1_640 = model1.val(
    data="/content/sds_swimmer/data.yaml",
    split="val",
    imgsz=640
)
```

```python
metrics4_640 = model4.val(
    data="/content/sds_swimmer/data.yaml",
    split="val",
    imgsz=640
)
```

Validation at 800

```python
metrics1_800 = model1.val(
    data="/content/sds_swimmer/data.yaml",
    split="val",
    imgsz=800
)
```

```python
metrics4_800 = model4.val(
    data="/content/sds_swimmer/data.yaml",
    split="val",
    imgsz=800
)
```

```python
import pandas as pd

comparison = pd.DataFrame({
    "Model": ["Model 1 - 640", "Model 4 - 640",
              "Model 1 - 800", "Model 4 - 800"],

    "Precision": [metrics1_640.box.mp, metrics4_640.box.mp,
                  metrics1_800.box.mp, metrics4_800.box.mp],

    "Recall": [metrics1_640.box.mr, metrics4_640.box.mr,
               metrics1_800.box.mr, metrics4_800.box.mr],

    "mAP50": [metrics1_640.box.map50, metrics4_640.box.map50,
              metrics1_800.box.map50, metrics4_800.box.map50],

    "mAP50-95": [metrics1_640.box.map, metrics4_640.box.map,
                 metrics1_800.box.map, metrics4_800.box.map]
})

comparison.round(3)
```
